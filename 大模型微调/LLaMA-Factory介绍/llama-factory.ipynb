## 界面参数说明
1. 训练轮数：至少是300轮以上 ， 模型才会被训练的比较稳定
2. 最大梯度范数 1.0 默认就行
3. 最大样本数： 默认就行，不能太小如果 设置成100，但是数据集有 1000条，那么就只会获取100条来训练
4. 计算类型： bf16 默认就行，千问1.5 就是bf16 的精度，开启 bf32 没有意义
5. 截断长度： 这个字段要根据数据集的一行的数据来给，这个数据长度就和显存有关，越大，显存使用越多， 训练的时候，一行文本的长度，maxlength
6. 批处理大小： 这个就根据显存条，先给小点，让集群的显存跑到90%即可
7. 梯度累积： 不用管 【和优化器相关的】
8. 验证集比例：0.1 ，这个根据数据集的大小来看着给，也可以不给
9. 学习率调节器： 不用管

## 预览命令 保存训练参数 载入训练参数 开始 中断
输出目录： 训练的参数保存路径
1. 配置路径： 也就是上面配置的参数保存的文件路径
2. 设备数量： 当前设置上有几张卡
3. deepspeed stage ： 当设备大于1 的时候，就可以使用多卡训练了
## 其他参数设置
1. 日志间隔： 5 ，也就是每5个**EPOCH** 打印一次
2. 保存间隔： 100 ，100个 **EPOCH** 保存一次 参数 [保存完成了以后会进行验证 ， 如何设置了验证的话]
