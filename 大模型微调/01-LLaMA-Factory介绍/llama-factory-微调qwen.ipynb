## 界面参数说明
1. 训练轮数：至少是300轮以上 ， 模型才会被训练的比较稳定
2. 最大梯度范数 1.0 默认就行
3. 最大样本数： 默认就行，不能太小如果 设置成100，但是数据集有 1000条，那么就只会获取100条来训练
4. 计算类型： bf16 默认就行，千问1.5 就是bf16 的精度，开启 bf32 没有意义
5. 截断长度： 这个字段要根据数据集的一行的数据来给，这个数据长度就和显存有关，越大，显存使用越多， 训练的时候，一行文本的长度，maxlength
6. 批处理大小： 这个就根据显存条，先给小点，让集群的显存跑到90%即可
7. 梯度累积： 不用管 【和优化器相关的】
8. 验证集比例：0.1 ，这个根据数据集的大小来看着给，也可以不给
9. 学习率调节器： 不用管

## 什么时候可以训练停下来了
模型已经收敛了，什么是收敛，也就是loss 的损失趋于平缓不动了，也就是躺平，这个时候模型应该是训练的比较好了

## 量化参数
1. 量化等级： 8位 
2. 量化方法： 不管
3. 对话模板： qwen ，使用的是千问模型
4. rope差值方法: none
5. 加速方式： flashattn2  ，训练过程中会将中间的特种做一个缓存，降低了重复计算的次数，所以整体的训练速度会快些，但是占用的显存会多一些

### 说明
1. 在模型微调训练过程中我们可以采用降低模型参数的方法来减少模型参数精度来节约训练的显存，从而提高训练批次大小，使的模型训练的更快
### 问题 : 之前的部署量量化有提到，量化是以牺牲精度为代价提升性能，那么在训练过程中使用量化是否会降低模型的精度呢？
回答: 1. 不会. 在量化部署中，由于模型的参数已经固定了，所以降低精度一定会影响结果，但是，在训练的过程中，模型的参数并未固定，依然处于学习的状态，所以降低参数的精度对模型的结果不会有太大的影响（应为AI训练的结果不是具体的值，而是一种趋势）。
2. 注意： 在量化微调中，量化只发生在内部的训练过程，并不影响模型最终的数据类型（模型原有的参数类型是f16，在量化微调训练中会量化为8位，参数保存时又会还原到f16，因此，量化微调并不影响模 型的最终数据类型）。
